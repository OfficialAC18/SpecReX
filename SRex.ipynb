```python
import os
import numpy as np
from tqdm import tqdm 
import onnxruntime as ort
import matplotlib.pyplot as plt
```

# Running SpecReX
1. First, import SRex from SpecReX.wrapper
2. Intialise an SReX object, by passing it a model (Can be ONNX, PyTorch or a prediction function)
3. Call the calc_responsibility function to calculate the responsibility landscape, need to pass spectra and wavenumber as arguments, other arguments are optional
4. To extract explanations, call the explanations() function to obtain wavenumber coords and beam widths (Required for visualisation)
5. To Visualise results, call visualise() function after calculating responsibility and extracting explanations

```python
#Testing SpecReX
from SpecReX.wrapper import SReX
```

```python
DIR = 'Null Simulations/Double Peaks'
```

```python
all_spectra = [os.path.join(DIR,file) for file in os.listdir(DIR) if 'class' in file and 'npy' in file]
all_models =  [os.path.join(DIR,file) for file in os.listdir(DIR) if '.onnx' in file]
wn = np.load(os.path.join(DIR,'wn.npy')).reshape(1,-1)
```

```python
#Go through all models and examples and save output to disk
for model_name in tqdm(all_models):
    #Load a model (Can be either ONNX or PyTorch)
    #Code below for loading ONNX model
    sess_opt = ort.SessionOptions()
    sess_opt.intra_op_num_threads = 0 #Just to supress unimportant warnings
    providers = ['CPUExecutionProvider'] #Set to [("CUDAExecutionProvider", {"enable_cuda_graph": False})] for GPU

    model = ort.InferenceSession(model_name, sess_options = sess_opt,
                             providers = providers)

    #Intialise the srex object
    srex = SReX(model = model)

    for spectra_name in all_spectra:
        #Calculate the responsibility
        spectra = np.load(spectra_name)
        resp = srex.calc_responsibility(spectra.reshape(1,-1),
                                        wn)
        resp = resp[0]


        # Plot the spectrum with clear distinction between high and low values
        fig, ax = plt.subplots(figsize=(12, 6))
        im = ax.imshow(resp[np.newaxis, :], aspect='auto', cmap='viridis',
                        extent=[0, len(spectra.reshape(-1)), np.min(spectra), np.max(spectra)])

        # Set axis labels and title
        ax.set_xlabel('Wavenumber Shift (cm$^{-1}$)')
        ax.set_ylabel('Responsibility (A.U.)')
        ax.set_title('Raman Spectra')

        #Also plot the spectra over it
        ax.plot(spectra, color = 'black')

        # Add a colorbar
        cbar = fig.colorbar(im, ax=ax)
        cbar.set_label('Responsibility (A.U.)')

        #Save to drive
        if 'NULL' in model_name:
            output_path = os.path.join(DIR,'Outputs Nathan CNN',
                                       'output_'+str(srex.targets)+'_'+spectra_name.split('/')[2][:-4]+'.jpg')
        
        elif 'lrcn' in model_name:
            output_path = os.path.join(DIR,'Outputs LRCN',
                                       'output_'+str(srex.targets)+'_'+spectra_name.split('/')[2][:-4]+'.jpg')

        elif 'conv' in model_name:
            output_path = os.path.join(DIR,'Outputs CNN',
                            'output_'+str(srex.targets)+'_'+spectra_name.split('/')[2][:-4]+'.jpg')

        elif 'lstm' in model_name:
            output_path = os.path.join(DIR,'Outputs LSTM',
                            'output_'+str(srex.targets)+'_'+spectra_name.split('/')[2][:-4]+'.jpg')
            
        #Save to drive
        fig.savefig(
            output_path,
            dpi = 900,
            bbox_inches = "tight"
        )
        
        plt.close()
            
```
