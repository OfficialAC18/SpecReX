```python
import captum
import torch
import numpy as np
import onnxruntime as ort
from SpecReX.wrapper import SReX
```

```python
# Load the ONNX model
onnx_model_path = "Trained Models/double_peak_lrcn_null.onnx"
sess_options = ort.SessionOptions()
sess_options.intra_op_num_threads = 0
providers = [("CUDAExecutionProvider", {"enable_cuda_graph": False})]

model = ort.InferenceSession(onnx_model_path, sess_options,
                            providers=providers)

#Intialise SReX 
srex = SReX(model=model)
```

```python
#Load the relevant spectra
spectra = np.load('Simulated Dataset/Double Peak/test1236_class0.npy')

#Calculate responsibility
resp = srex.calc_responsibility(spectra.reshape(1,-1))
```
